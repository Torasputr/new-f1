```python
import os
from dotenv import load_dotenv
from google.cloud import storage
import pandas as pd
from io import BytesIO

load_dotenv()
```

```python
SEASON = int(os.getenv("SEASON", ""))

GCS_BUCKET = os.getenv("GCS_BUCKET", "").strip()
GCS_MEDAL = os.getenv("GCS_MEDAL", "").strip()
BLOB_PATH = f"{GCS_MEDAL}/season={SEASON}/session.parquet"
```

```python
print(f"[INFO] Fetching sessions for season {SEASON}")
client = storage.Client()
blob = client.bucket(GCS_BUCKET).blob(BLOB_PATH)

df = pd.read_parquet(BytesIO(blob.download_as_bytes()))
df.to_csv("../../data/silver/sessions.csv")
```

```python
df = df.drop(columns=["session_type"], errors="coerce")
```
